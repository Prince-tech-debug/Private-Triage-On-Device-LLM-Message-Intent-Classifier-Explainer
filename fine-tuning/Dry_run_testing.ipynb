```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_community.llms import LlamaCpp

```

```python
# from langchain_community.llms import LlamaCpp

# 1. Define your messages exact to your Colab setup
messages = [
    {
        "role": "system",
        "content": "You are a concise operational text summarizer. Provide a sharp, 1-sentence summary of the message."
    },
    {
        "role": "user",
        "content": "Internshala B.Tech students, explore this Work from Home opportunity Internshala Student Partner Program Are you looking for work opportunities in your free time? This program connects students with relevant opportunities. Program highlights: Learning Opportunity - Masterclasses by industry professionals Skill Development - Gain professional experience Professional Development - Develop skills like Leadership, Communication & Marketing What's more? Add program experience to your resume to boost it. Eligibility - B.Tech students from Madhav Institute can apply. Click here for more information. Applications are currently open"
    }
]

# 2. Replicate `apply_chat_template(add_generation_prompt=True)` for Gemma 3
def format_gemma_prompt(messages: list) -> str:
    system_text = ""
    user_text = ""
    
    for msg in messages:
        if msg["role"] == "system":
            # Extract string content whether it's a raw string or structured dict list
            if isinstance(msg["content"], list):
                system_text = msg["content"][0]["text"]
            else:
                system_text = msg["content"]
        elif msg["role"] == "user":
            if isinstance(msg["content"], list):
                user_text = msg["content"][0]["text"]
            else:
                user_text = msg["content"]
                
    # Gemma 3 combines system instructions into the first user block
    prompt = (
        f"<start_of_turn>user\n"
        f"{system_text}\n\n"
        f"{user_text}<end_of_turn>\n"
        f"<start_of_turn>model\n"
    )
    return prompt

# Format the input string
prompt_string = format_gemma_prompt(messages)

# 3. Initialize LlamaCpp with generation parameters matching Colab
llm = LlamaCpp(
    model_path="../my_model/gemma-3-4b-it.Q4_K_M.gguf",
    temperature=0.7,
    max_tokens=256,
    stop=["<end_of_turn>", "<eos>"], 
    verbose=False
    
)

# 4. Generate summary
response = llm.invoke(prompt_string)
print(response.strip())
```
