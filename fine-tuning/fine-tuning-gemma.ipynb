```python
#imports
import torch
from datasets import Dataset
import pandas as pd
from transformers import (
    AutoProcessor,
    AutoModelForCausalLM,
    BitsAndBytesConfig,
    TrainingArguments
)
from peft import LoraConfig, get_peft_model, prepare_model_for_kbit_training
from trl import SFTTrainer
```

```python
MODEL = "google/gemma-3-4b-it" # Now we have the model lets check if the hardware is supported by bfloat16
compute_dtype = torch.bfloat16 if torch.cuda.is_available() and torch.cuda.is_balance_supported() else torch.float16

bnb_config =BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype=compute_dtype,
    bnb_4bit_use_double_quant=True
)
```

```python
processor = AutoProcessor.from_pretrained(MODEL)
model = AutoModelForMultimodalLM.from_pretrained(
    MODEL,
    quantization_config = bnb_config,
    device_map = 'auto'
)
```
